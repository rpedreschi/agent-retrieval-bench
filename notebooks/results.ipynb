# agent-retrieval-bench: results notebook

Loads the JSONL produced by `arb eval` and renders every chart used in the blog
post and conference talks.

By default this notebook reads `tests/fixtures/example_runs.jsonl` so it works
without a real benchmark run. Point `RUNS_PATH` at the output of
`arb eval --execute --out runs/headline.jsonl` to render headline numbers.

Methodology recap (see `docs/methodology.md`):
- Snapshot-first grading.
- Per-agent MCP token scoping.
- Headline metric: `cost_per_correct_degraded`.

```python
from pathlib import Path
from arb.analysis import load_runs
from arb.analysis.charts import (
    chart_failure_taxonomy,
    chart_success_rate_by_condition,
    chart_cost_per_correct,
    chart_summary_table,
)

RUNS_PATH = Path('../tests/fixtures/example_runs.jsonl')
rows = load_runs(RUNS_PATH)
len(rows)
```

```python
import json
print(json.dumps(chart_summary_table(rows), indent=2, default=str))
```

## Failure taxonomy (headline chart)

Per-variant histogram over the failure categories. Variant A's bar should
show fat `missed_join` and `temporal_misalignment` bands; Variant B's bar
should be much smaller and differently shaped.

```python
fig, ax = chart_failure_taxonomy(rows)
fig
```

## Success rate by condition (with 95% CIs)

Clean / degraded / adversarial. The degraded panel matters most — that's
production.

```python
fig, ax = chart_success_rate_by_condition(rows)
fig
```

## Cost per correct task

Bootstrapped CI per variant per condition. The right-hand bar pair is the
`cost_per_correct_degraded` headline.

```python
fig, ax = chart_cost_per_correct(rows)
fig
```
